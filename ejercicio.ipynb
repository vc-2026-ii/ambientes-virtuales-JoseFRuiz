# Construcción de imágenes

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = np.array([[[255,255,255],[255,255,255],[255,255,255],[255,255,255],[255,255,255]],
                [[255,255,255],[0,255,0],[0,255,0],[0,255,0],[255,255,255]],
                [[255,255,255],[0,255,0],[255,255,255],[0,255,0],[255,255,255]],
                [[255,255,255],[0,255,0],[0,255,0],[0,255,0],[255,255,255]],
                [[255,255,255],[0,255,0],[255,255,255],[0,255,0],[255,255,255]],
                [[255,255,255],[255,255,255],[255,255,255],[255,255,255],[255,255,255]]])
```

```python
plt.figure()
plt.imshow(img)
```

```python
imgr = np.array([[255, 255, 255, 255, 255],
                 [255, 0, 0, 0, 255],
                 [255, 0, 255, 0, 255],
                 [255, 0, 0, 0, 255],
                 [255, 0, 255, 0, 255],
                 [255, 255, 255, 255, 255]])

imgg = np.array([[255, 255, 255, 255, 255],
                 [255, 255, 255, 255, 255],
                 [255, 255, 255, 255, 255],
                 [255, 255, 255, 255, 255],
                 [255, 255, 255, 255, 255],
                 [255, 255, 255, 255, 255]])

imgb = np.array([[255, 255, 255, 255, 255],
                 [255, 0, 0, 0, 255],
                 [255, 0, 255, 0, 255],
                 [255, 0, 0, 0, 255],
                 [255, 0, 255, 0, 255],
                 [255, 255, 255, 255, 255]])


imgs = np.stack([imgr, imgg, imgb], axis=2)
```

```python
imgs.shape
```

```python
plt.figure()
plt.imshow(imgs)
```

```python
img2 = 255*np.ones((6,5,3))
img2[1,1:4,:] = [0, 255, 0]
img2[1:5,1,:] = [0, 255, 0]
img2[1:5,3,:] = [0, 255, 0]
img2[3,2,:] = [0, 255, 0]

plt.figure()
plt.imshow(img2)
```

# Concatenación

```python
imgs2 = np.concatenate([imgs, imgs], axis=0)

plt.figure()
plt.imshow(imgs2)
```

```python
imgs2.shape
```

```python
imgs3 = np.concatenate([imgs, imgs], axis=1)
plt.figure()
plt.imshow(imgs3)
```

```python
imgs3.shape
```
